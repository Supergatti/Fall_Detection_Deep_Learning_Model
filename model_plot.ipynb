# 绘制 CNN 和 LSTM 模型结构图

本笔记本用于加载预训练的 CNN 和 LSTM 模型，并生成它们的结构可视化图。

```python
# 安装必要的依赖
%pip install tensorflow pydot graphviz
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.utils import plot_model
import os

# 确认模型文件路径
lstm_model_path = 'Trained_model/final_lstm_model.h5'
print(f"LSTM 模型文件是否存在: {os.path.exists(lstm_model_path)}")

try:
    # 加载 LSTM 模型
    lstm_model = keras.models.load_model(lstm_model_path)
    print("LSTM 模型加载成功!")
    
    # 打印 LSTM 模型摘要
    lstm_model.summary()
    
    # 绘制并保存 LSTM 模型结构图
    plot_model(lstm_model, show_shapes=True, to_file='lstm_model.png')
    print("LSTM 模型结构图已保存为 'lstm_model.png'")
except Exception as e:
    print(f"加载 LSTM 模型时出错: {e}")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.utils import plot_model
import os

# 确认模型文件路径
cnn_model_path = 'Trained_model/final_cnn_model.h5'
print(f"CNN 模型文件是否存在: {os.path.exists(cnn_model_path)}")

try:
    # 加载 CNN 模型
    cnn_model = keras.models.load_model(cnn_model_path)
    print("CNN 模型加载成功!")
    
    # 打印 CNN 模型摘要
    cnn_model.summary()
    
    # 绘制并保存 CNN 模型结构图
    plot_model(cnn_model, show_shapes=True, to_file='cnn_model.png')
    print("CNN 模型结构图已保存为 'cnn_model.png'")
except Exception as e:
    print(f"加载 CNN 模型时出错: {e}")
```

## 如果上述加载失败，尝试手动重建模型结构

如果由于版本兼容性问题无法加载预训练模型，我们可以根据模型结构手动重建它们，然后绘制结构图。

```python
# 手动构建 LSTM 模型结构
def create_lstm_model():
    model = keras.Sequential()
    model.add(keras.layers.LSTM(64, return_sequences=True, input_shape=(30, 42)))
    model.add(keras.layers.LSTM(32))
    model.add(keras.layers.Dense(16, activation='relu'))
    model.add(keras.layers.Dense(1, activation='sigmoid'))
    return model

# 创建并绘制 LSTM 模型
try:
    lstm_model = create_lstm_model()
    plot_model(lstm_model, show_shapes=True, to_file='lstm_model_manual.png')
    print("手动构建的 LSTM 模型结构图已保存为 'lstm_model_manual.png'")
except Exception as e:
    print(f"绘制手动构建的 LSTM 模型时出错: {e}")
```

```python
# 手动构建 CNN 模型结构
def create_cnn_model():
    model = keras.Sequential()
    model.add(keras.layers.Conv2D(64, (3, 3), activation='relu', input_shape=(30, 42, 1)))
    model.add(keras.layers.MaxPooling2D((2, 2)))
    model.add(keras.layers.Conv2D(32, (3, 3), activation='relu'))
    model.add(keras.layers.MaxPooling2D((2, 2)))
    model.add(keras.layers.Flatten())
    model.add(keras.layers.Dense(16, activation='relu'))
    model.add(keras.layers.Dense(1, activation='sigmoid'))
    return model

# 创建并绘制 CNN 模型
try:
    cnn_model = create_cnn_model()
    plot_model(cnn_model, show_shapes=True, to_file='cnn_model_manual.png')
    print("手动构建的 CNN 模型结构图已保存为 'cnn_model_manual.png'")
except Exception as e:
    print(f"绘制手动构建的 CNN 模型时出错: {e}")
```
